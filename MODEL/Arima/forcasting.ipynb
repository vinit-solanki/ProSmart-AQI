```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import seaborn as sns
from datetime import datetime
import json

# Read the data
df = pd.read_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/DataPreprocessing/datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)

def analyze_acf_pacf(series, title):
    """Analyze ACF and PACF plots for a time series"""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    
    # Plot ACF
    plot_acf(series.dropna(), ax=ax1, lags=40)
    ax1.set_title(f'Autocorrelation Function for {title}')
    
    # Plot PACF
    plot_pacf(series.dropna(), ax=ax2, lags=40)
    ax2.set_title(f'Partial Autocorrelation Function for {title}')
    
    plt.tight_layout()
    plt.show()

def suggest_order(series, title):
    """Suggest ARIMA order based on ACF/PACF analysis"""
    # Check if differencing is needed
    result = adfuller(series.dropna())
    d = 1 if result[1] > 0.05 else 0
    
    # Store ADF test results
    adf_statistic = result[0]
    p_value = result[1]
    critical_values = result[4]
    
    # Print ADF test results for original series
    print("\nADF Test Results (Original Series):")
    print(f"ADF Statistic: {adf_statistic:.4f}")
    print(f"p-value: {p_value:.4f}")
    print("Critical Values:")
    for key, value in critical_values.items():
        print(f"\t{key}: {value:.4f}")
    
    # If d=1, analyze the differenced series
    if d == 1:
        series_diff = series.diff().dropna()
        result_diff = adfuller(series_diff.dropna())
        
        # Print ADF test results for differenced series
        print("\nADF Test Results (After Differencing):")
        print(f"ADF Statistic: {result_diff[0]:.4f}")
        print(f"p-value: {result_diff[1]:.4f}")
        print("Critical Values:")
        for key, value in result_diff[4].items():
            print(f"\t{key}: {value:.4f}")
        
        series = series_diff
    
    # Plot ACF/PACF for the (differenced) series
    print(f"\nACF and PACF plots for {title}")
    analyze_acf_pacf(series, title)
    
    return d, series

def estimate_arima_orders(series):
    """Estimate p and q orders based on ACF and PACF values"""
    n = len(series)
    
    # Calculate ACF and PACF values
    acf_values = acf(series.dropna(), nlags=40, fft=False)
    pacf_values = pacf(series.dropna(), nlags=40)
    
    # Calculate confidence interval
    confidence_interval = 1.96 / np.sqrt(n)
    
    # Estimate p (AR order) from PACF
    p = 0
    consecutive_insignificant = 0
    for i in range(1, len(pacf_values)):
        if abs(pacf_values[i]) > confidence_interval:
            p = i
            consecutive_insignificant = 0
        else:
            consecutive_insignificant += 1
            if consecutive_insignificant >= 2 and i > 2:
                break
    
    # Estimate q (MA order) from ACF
    q = 0
    consecutive_insignificant = 0
    for i in range(1, len(acf_values)):
        if abs(acf_values[i]) > confidence_interval:
            q = i
            consecutive_insignificant = 0
        else:
            consecutive_insignificant += 1
            if consecutive_insignificant >= 2 and i > 2:
                break
    
    # Limit maximum orders
    p = min(p, 5)
    q = min(q, 5)
    
    # Ensure minimum orders
    p = max(2, p)
    q = max(2, q)
    
    return p, q, acf_values, pacf_values

def plot_model_diagnostics(acf_values, pacf_values, title):
    """Plot ACF and PACF with significance levels"""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    
    # Plot ACF
    lags = range(len(acf_values))
    ax1.vlines(lags, [0], acf_values)
    ax1.axhline(y=0, color='black', linestyle='-')
    ax1.axhline(y=1.96/np.sqrt(len(acf_values)), color='r', linestyle='--', label='95% Confidence Interval')
    ax1.axhline(y=-1.96/np.sqrt(len(acf_values)), color='r', linestyle='--')
    ax1.set_title(f'ACF for {title}')
    ax1.set_xlabel('Lag')
    ax1.set_ylabel('Correlation')
    ax1.legend()
    
    # Plot PACF
    ax2.vlines(lags, [0], pacf_values)
    ax2.axhline(y=0, color='black', linestyle='-')
    ax2.axhline(y=1.96/np.sqrt(len(pacf_values)), color='r', linestyle='--', label='95% Confidence Interval')
    ax2.axhline(y=-1.96/np.sqrt(len(pacf_values)), color='r', linestyle='--')
    ax2.set_title(f'PACF for {title}')
    ax2.set_xlabel('Lag')
    ax2.set_ylabel('Partial Correlation')
    ax2.legend()
    
    plt.tight_layout()
    plt.show()

# Dictionary to store ARIMA parameters
arima_parameters = {}

# List of pollutants to analyze
pollutants = ['PM2.5_filled', 'PM10_filled', 'NO_filled', 'NO2_filled', 
              'NOx_filled', 'NH3_filled', 'Ozone_filled', 'Benzene_filled']

# Analyze each pollutant
for pollutant in pollutants:
    print(f"\nAnalyzing {pollutant}")
    print("=" * 50)
    
    # Get differencing order and transformed series
    d, transformed_series = suggest_order(df[pollutant], pollutant)
    
    # Estimate ARIMA orders and get ACF/PACF values
    p, q, acf_values, pacf_values = estimate_arima_orders(transformed_series)
    
    # Plot diagnostic plots with significance levels
    plot_model_diagnostics(acf_values, pacf_values, pollutant)
    
    print(f"\nSuggested ARIMA({p},{d},{q}) parameters for {pollutant}:")
    print(f"- p (AR order): {p}")
    print(f"- d (Differencing): {d}")
    print(f"- q (MA order): {q}")
    print("-" * 50)
    
    # Store parameters in dictionary
    arima_parameters[pollutant] = {
        'p': p,
        'd': d,
        'q': q
    }

# Convert dictionary to DataFrame
params_df = pd.DataFrame.from_dict(arima_parameters, orient='index')
print("\nARIMA Parameters Summary:")
print(params_df)

# Save parameters to CSV
params_df.to_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/arima_parameters.csv')

# Save parameters to JSON
with open('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/arima_parameters.json', 'w') as f:
    json.dump(arima_parameters, f, indent=4)

print("\nParameters have been saved to:")
print("- arima_parameters.csv")
print("- arima_parameters.json")

# Function to load parameters in other scripts
def load_arima_parameters():
    """Load ARIMA parameters from JSON file"""
    with open('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/arima_parameters.json', 'r') as f:
        return json.load(f)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller, acf, pacf
import json
import warnings
warnings.filterwarnings('ignore')

# Load the data
df = pd.read_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/DataPreprocessing/datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)

def check_stationarity(series):
    """Check stationarity using Augmented Dickey-Fuller test"""
    result = adfuller(series)
    return result[1] < 0.05  # Returns True if stationary

def determine_d(series):
    """Determine the order of differencing (d)"""
    d = 0
    while not check_stationarity(series):
        series = series.diff().dropna()
        d += 1
        if d > 2:  # Usually we don't need d > 2
            break
    return d

def find_p_q(series):
    """Automatically find p and q values using ACF and PACF"""
    # Calculate ACF and PACF
    acf_values = acf(series, nlags=40)
    pacf_values = pacf(series, nlags=40)
    
    # Calculate confidence intervals (95%)
    confidence_interval = 1.96 / np.sqrt(len(series))
    
    # Find p (AR order) from PACF
    p = 0
    for i in range(1, len(pacf_values)):
        if abs(pacf_values[i]) > confidence_interval:
            p = i
    
    # Find q (MA order) from ACF
    q = 0
    for i in range(1, len(acf_values)):
        if abs(acf_values[i]) > confidence_interval:
            q = i
    
    return p, q

def analyze_acf_pacf(series, title):
    """Plot ACF and PACF"""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    
    plot_acf(series, ax=ax1, lags=40)
    ax1.set_title(f'ACF Plot for {title}')
    
    plot_pacf(series, ax=ax2, lags=40)
    ax2.set_title(f'PACF Plot for {title}')
    
    plt.tight_layout()
    plt.show()

def find_parameters(series, title):
    """Find ARIMA parameters automatically"""
    print(f"\nAnalyzing {title}")
    
    # Find d (differencing order)
    d = determine_d(series)
    print(f"Differencing order (d): {d}")
    
    # If d > 0, analyze the differenced series
    if d > 0:
        series_diff = series.diff(d).dropna()
    else:
        series_diff = series
    
    # Plot ACF and PACF
    analyze_acf_pacf(series_diff, title)
    
    # Find p and q automatically
    p, q = find_p_q(series_diff)
    print(f"AR order (p): {p}")
    print(f"MA order (q): {q}")
    
    return {'p': p, 'd': d, 'q': q}

# Dictionary to store ARIMA parameters
arima_parameters = {}

# Find parameters for each pollutant
pollutants = ['PM2.5_filled', 'PM10_filled', 'NO_filled', 'NO2_filled', 
              'NOx_filled', 'NH3_filled', 'Ozone_filled', 'Benzene_filled']

for pollutant in pollutants:
    arima_parameters[pollutant] = find_parameters(df[pollutant], pollutant)

# Save parameters to JSON
with open('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/arima_parameters.json', 'w') as f:
    json.dump(arima_parameters, f, indent=4)

print("\nParameters have been saved to 'arima_parameters.json'")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from itertools import product
import json
import warnings
warnings.filterwarnings('ignore')

# Load the data
df = pd.read_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/DataPreprocessing/datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)

def check_stationarity(series, alpha=0.05):
    """Enhanced stationarity check with more metrics"""
    result = adfuller(series)
    p_value = result[1]
    is_stationary = p_value < alpha
    
    # Additional checks
    if is_stationary:
        # Check if the series has constant mean and variance visually
        rolling_mean = series.rolling(window=12).mean()
        rolling_std = series.rolling(window=12).std()
        
        plt.figure(figsize=(12, 5))
        plt.plot(series, label='Original')
        plt.plot(rolling_mean, label='Rolling Mean')
        plt.plot(rolling_std, label='Rolling Std')
        plt.legend()
        plt.title(f'Rolling Mean & Standard Deviation (p={p_value:.4f})')
        plt.show()
    
    return is_stationary

def determine_d(series, max_d=2):
    """More robust differencing order determination"""
    d = 0
    current_series = series.copy()
    
    while not check_stationarity(current_series) and d <= max_d:
        d += 1
        current_series = current_series.diff().dropna()
        
        # Plot each differenced series for visual inspection
        plt.figure(figsize=(12, 4))
        plt.plot(current_series)
        plt.title(f'{d} order differencing')
        plt.show()
    
    return d

def find_optimal_pq(series, max_lag=10):
    """Improved p,q selection using information criteria"""
    best_aic = np.inf
    best_bic = np.inf
    best_pq = (0, 0)
    
    # Try different combinations
    for p in range(0, max_lag+1):
        for q in range(0, max_lag+1):
            try:
                model = ARIMA(series, order=(p, 0, q))
                results = model.fit()
                
                if results.aic < best_aic:
                    best_aic = results.aic
                    best_pq = (p, q)
            except:
                continue
    
    return best_pq

def grid_search_arima(series, p_range, d_range, q_range):
    """Comprehensive grid search for best ARIMA parameters"""
    best_score = np.inf
    best_order = None
    best_model = None
    
    for order in product(p_range, d_range, q_range):
        try:
            model = ARIMA(series, order=order)
            results = model.fit()
            
            # Use BIC as it penalizes complex models more than AIC
            if results.bic < best_score:
                best_score = results.bic
                best_order = order
                best_model = results
                
        except Exception as e:
            print(f"Failed for order {order}: {str(e)}")
            continue
            
    return best_order, best_model



def evaluate_arima_model(series, order, train_size=0.8):
    """Evaluate ARIMA model with walk-forward validation"""
    # Split data
    n_obs = int(len(series) * train_size)
    train, test = series[:n_obs], series[n_obs:]
    
    # Walk-forward validation
    history = [x for x in train]
    predictions = []
    
    for t in range(len(test)):
        model = ARIMA(history, order=order)
        model_fit = model.fit()
        yhat = model_fit.forecast()[0]
        predictions.append(yhat)
        history.append(test[t])
    
    # Calculate metrics
    rmse = np.sqrt(mean_squared_error(test, predictions))
    mae = mean_absolute_error(test, predictions)
    r2 = r2_score(test, predictions)
    
    # Plot results
    plt.figure(figsize=(12, 6))
    # Convert to numpy arrays for plotting
    test_values = np.array(test)
    pred_values = np.array(predictions)
    dates = test.index.values
    
    plt.plot(dates, test_values, label='Actual', color='blue')
    plt.plot(dates, pred_values, color='red', label='Predicted')
    plt.title(f'ARIMA {order} Forecast\nRMSE: {rmse:.2f}, MAE: {mae:.2f}, R²: {r2:.2f}')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return rmse, mae, r2

def make_future_predictions(series, order, steps=30):
    """Make future predictions using the ARIMA model"""
    model = ARIMA(series, order=order)
    model_fit = model.fit()
    
    forecast = model_fit.forecast(steps=steps)
    last_date = series.index[-1]
    future_dates = pd.date_range(start=last_date + pd.Timedelta(days=1), periods=steps, freq='D')
    
    pred_conf = model_fit.get_forecast(steps=steps)
    conf_int = pred_conf.conf_int()
    
    plt.figure(figsize=(15, 7))
    # Convert series to numpy arrays before plotting
    historical_dates = series.index.to_numpy()
    historical_values = series.to_numpy()
    plt.plot(historical_dates, historical_values, label='Historical Data', color='blue')
    plt.plot(future_dates, forecast, color='red', label='Forecast')
    plt.fill_between(future_dates, 
                    conf_int.iloc[:, 0], 
                    conf_int.iloc[:, 1], 
                    color='red', alpha=0.2, 
                    label='95% Confidence Interval')
    plt.title(f'ARIMA{order} Forecast for next {steps} days')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return pd.Series(forecast, index=future_dates), conf_int

def find_best_parameters(series, title):
    """Comprehensive parameter finding with evaluation"""
    print(f"\n=== Analyzing {title} ===")
    
    # Step 1: Determine differencing order
    d = determine_d(series)
    print(f"Suggested differencing order (d): {d}")
    
    # Step 2: Get stationary series
    if d > 0:
        stationary_series = series.diff(d).dropna()
    else:
        stationary_series = series.copy()
    
    # Step 3: Plot ACF/PACF for visual inspection
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    plot_acf(stationary_series, ax=ax1, lags=40)
    plot_pacf(stationary_series, ax=ax2, lags=40)
    plt.suptitle(f'ACF/PACF for {title} (d={d})')
    plt.tight_layout()
    plt.show()
    
    # Step 4: Grid search for best parameters
    p_range = range(0, 3)  # Try 0-2 for AR
    q_range = range(0, 3)  # Try 0-2 for MA
    best_order, best_model = grid_search_arima(series, p_range, [d], q_range)
    
    if best_order is None:
        print("No suitable model found with grid search")
        return None
    
    print(f"Best ARIMA order from grid search: {best_order}")
    print(best_model.summary())
    
    # Step 5: Evaluate the model
    rmse, mae, r2 = evaluate_arima_model(series, best_order)
    
    # Step 6: Return parameters and performance
    return {
        'p': best_order[0],
        'd': best_order[1],
        'q': best_order[2],
        'rmse': rmse,
        'mae': mae,
        'r2': r2,
        'aic': best_model.aic,
        'bic': best_model.bic
    }

# Dictionary to store ARIMA parameters and performance
arima_results = {}

# Find parameters for each pollutant
pollutants = ['PM2.5_filled', 'PM10_filled', 'NO_filled', 'NO2_filled', 
              'NOx_filled', 'NH3_filled', 'Ozone_filled', 'Benzene_filled']
print("\n=== Making Future Predictions ===")
future_predictions = {}
for pollutant in pollutants:
    result = find_best_parameters(df[pollutant], pollutant)
    if result is not None:
        arima_results[pollutant] = result
for pollutant, params in arima_results.items():
    print(f"\nMaking predictions for {pollutant}")
    order = (params['p'], params['d'], params['q'])
    predictions, conf_intervals = make_future_predictions(df[pollutant], order)
    
    future_predictions[pollutant] = {
        'predictions': predictions.to_dict(),
        'lower_bound': conf_intervals.iloc[:, 0].to_dict(),
        'upper_bound': conf_intervals.iloc[:, 1].to_dict()
    }
# Save results to JSON
with open('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/future_predictions.json', 'w') as f:
    json.dump(future_predictions, f, indent=4)

print("\nFuture predictions saved to 'future_predictions.json'")
# Print summary of results
print("\n=== Prediction Summary ===")
for pollutant in future_predictions:
    pred_series = pd.Series(future_predictions[pollutant]['predictions'])
    print(f"\n{pollutant}:")
    print(f"Mean predicted value: {pred_series.mean():.2f}")
    print(f"Min predicted value: {pred_series.min():.2f}")
    print(f"Max predicted value: {pred_series.max():.2f}")
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import acf, pacf
import matplotlib.pyplot as plt

# Read and prepare data
df = pd.read_csv('./datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)
data = df['PM2.5_filled']

# Transform to stationary
datasetLogDiffShifting = np.log(data).diff().dropna()

# Calculate ACF/PACF
nlags = 10  # Extended to check for seasonality
lag_acf = acf(datasetLogDiffShifting, nlags=nlags)
lag_pacf = pacf(datasetLogDiffShifting, nlags=nlags, method='ols')

# Determine AR(P) and MA(Q) orders
Q = np.argmax(lag_acf < (1.96 / np.sqrt(len(datasetLogDiffShifting))))  # MA order
P = np.argmax(lag_pacf < (1.96 / np.sqrt(len(datasetLogDiffShifting))))  # AR order

# Plotting
plt.figure(figsize=(14, 6))

# ACF Plot
plt.subplot(121)
plt.plot(lag_acf, marker='o', linestyle='-', color='b')
plt.axhline(y=0, linestyle='--', color='gray')
plt.axhline(y=-1.96/np.sqrt(len(datasetLogDiffShifting)), linestyle='--', color='gray')
plt.axhline(y=1.96/np.sqrt(len(datasetLogDiffShifting)), linestyle='--', color='gray')
plt.title(f'ACF: PM2.5 (Potential MA({Q}))')

# PACF Plot
plt.subplot(122)
plt.plot(lag_pacf, marker='o', linestyle='-', color='r')
plt.axhline(y=0, linestyle='--', color='gray')
plt.axhline(y=-1.96/np.sqrt(len(datasetLogDiffShifting)), linestyle='--', color='gray')
plt.axhline(y=1.96/np.sqrt(len(datasetLogDiffShifting)), linestyle='--', color='gray')
plt.title(f'PACF: PM2.5 (Potential AR({P}))')

plt.tight_layout()
plt.show()

print(f"From the ACF graph, the curve touches y=0 at x={Q}. Thus, from theory, Q = {Q}")
print(f"From the PACF graph, the curve touches y=0 at x={P}. Thus, from theory, P = {P}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error
import scipy.stats as stats
from statsmodels.stats.diagnostic import acorr_ljungbox
# Load and prepare data
df = pd.read_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)
data = df['PM2.5_filled']

# Enhanced preprocessing with log1p
def preprocess_data(series):
    return np.log1p(series), 0  # Using log1p for better handling of zeros

data_transformed, _ = preprocess_data(data)
train_size = int(len(data_transformed) * 0.7)
train, test = data_transformed.iloc[:train_size], data_transformed.iloc[train_size:]

# ======================
# MODEL VALIDATION FRAMEWORK
# ======================

def walk_forward_validation(data, train_size, order, seasonal_order, steps=7, verbose=True):
    """Enhanced validation with rolling origin"""
    test = data.iloc[train_size:]
    n_windows = len(test) // steps
    metrics = []
    forecasts = []
    
    for i in range(n_windows):
        train_window = data.iloc[:train_size + i*steps]
        
        try:
            model = SARIMAX(train_window, 
                          order=order,
                          seasonal_order=seasonal_order,
                          enforce_stationarity=False,
                          enforce_invertibility=False)
            
            model_fit = model.fit(disp=False, maxiter=200)
            
            forecast = model_fit.get_forecast(steps=steps)
            pred = np.expm1(forecast.predicted_mean)
            actual = np.expm1(test.iloc[i*steps:(i+1)*steps])
            
            mae = mean_absolute_error(actual, pred)
            mape = np.mean(np.abs((actual - pred)/actual)) * 100
            metrics.append({
                'Window': i+1,
                'MAE': mae,
                'MAPE': mape,
                'AIC': model_fit.aic,
                'BIC': model_fit.bic
            })
            
            forecasts.append({
                'date': test.index[i*steps:(i+1)*steps],
                'actual': actual.values,
                'predicted': pred.values
            })
            
        except Exception as e:
            if verbose:
                print(f"Window {i+1} failed: {str(e)}")
            continue
            
    return pd.DataFrame(metrics), pd.concat([pd.DataFrame(f) for f in forecasts])

# ======================
# MODEL COMPARISON
# ======================

models_to_test = {
    "Original (2,1,2)x(1,1,1,30)": ((2,1,2), (1,1,1,30)),
    "Simplified (2,1,0)x(0,1,0,30)": ((2,1,0), (0,1,0,30)),
    "Robust (1,1,1)x(0,1,1,30)": ((1,1,1), (0,1,1,30)),
    "Auto (1,1,0)x(0,1,1,30)": ((1,1,0), (0,1,1,30))
}

validation_results = {}
forecast_data = {}

for name, params in models_to_test.items():
    print(f"\n=== Validating {name} ===")
    res, forecasts = walk_forward_validation(data_transformed, train_size, *params)
    validation_results[name] = res
    forecast_data[name] = forecasts
    print(res[['MAE', 'MAPE']].describe())

# ======================
# SELECT BEST MODEL
# ======================

model_scores = {name: res['MAE'].median() for name, res in validation_results.items()}
best_model_name = min(model_scores, key=model_scores.get)
best_order, best_seasonal = models_to_test[best_model_name]

print(f"\n★★★ Best Model: {best_model_name} ★★★")
print(f"Median MAE: {model_scores[best_model_name]:.2f}")
print(f"Order: {best_order}, Seasonal Order: {best_seasonal}")

# ======================
# FINAL MODEL TRAINING
# ======================

final_model = SARIMAX(data_transformed,
                     order=best_order,
                     seasonal_order=best_seasonal,
                     enforce_stationarity=False,
                     enforce_invertibility=False)

final_fit = final_model.fit(disp=False, maxiter=200)

def generate_forecast(model_fit, steps=30, history=None):
    """Generate production forecast with diagnostics"""
    # Get forecast
    forecast = model_fit.get_forecast(steps=steps)
    pred = pd.Series(np.expm1(forecast.predicted_mean), name='forecast')
    conf_int = np.expm1(forecast.conf_int())
    
    # Create output dataframe
    forecast_dates = pd.date_range(start=data.index[-1] + pd.Timedelta(days=1), periods=steps)
    result = pd.DataFrame({
        'date': forecast_dates,
        'PM2.5_forecast': pred.values,
        'lower_CI': conf_int.iloc[:,0],
        'upper_CI': conf_int.iloc[:,1]
    }).set_index('date')
    
    # Plot with history if provided
    plt.figure(figsize=(15,6))
    if history is not None:
        # Convert history to numpy array for plotting - FIXED
        history_dates = history.index[-90:].to_numpy()  # Convert to numpy array
        history_values = history.values[-90:]
        plt.plot(history_dates, history_values, 
                label='Historical PM2.5', alpha=0.7)
    
    # Convert result index to numpy array as well
    forecast_dates = result.index.to_numpy()
    forecast_values = result['PM2.5_forecast'].to_numpy()
    lower_ci = result['lower_CI'].to_numpy()
    upper_ci = result['upper_CI'].to_numpy()
    
    plt.plot(forecast_dates, forecast_values, 'r--', label='Forecast')
    plt.fill_between(forecast_dates, lower_ci, upper_ci, 
                    color='pink', alpha=0.3, label='95% CI')
    
    plt.title(f'PM2.5 {steps}-Day Forecast | Best Model: {best_model_name}')
    plt.ylabel('PM2.5 (μg/m³)')
    plt.legend()
    plt.grid()
    plt.show()
    
    return result


# Generate 30-day forecast (now with fixed plotting)
production_forecast = generate_forecast(final_fit, steps=30, history=data)

# ======================
# PRODUCTION READINESS REPORT
# ======================

def production_report(model_fit, validation_metrics):
    """Comprehensive deployment checklist"""
    print("\n" + "="*50)
    print("PRODUCTION READINESS REPORT".center(50))
    print("="*50)
    
    # 1. Model Parameters
    print("\n1. MODEL PARAMETERS:")
    print(f"SARIMAX{model_fit.model.order}x{model_fit.model.seasonal_order}")
    print("\nSignificant parameters (p < 0.05):")
    sig_params = model_fit.pvalues[model_fit.pvalues < 0.05]
    print(sig_params.to_string())
    
    # 2. Residual Analysis
    print("\n2. RESIDUAL DIAGNOSTICS:")
    residuals = model_fit.resid
    jb_test = stats.jarque_bera(residuals)
    # Use the imported acorr_ljungbox function
    lb_test = acorr_ljungbox(residuals, lags=10)
    
    print(f"- Normality (JB test): p-value = {jb_test[1]:.4f}")
    print(f"- Autocorrelation (Ljung-Box): p-value = {lb_test.iloc[-1,1]:.4f}")
    
    # 3. Validation Performance
    print("\n3. VALIDATION PERFORMANCE:")
    print(validation_metrics[['MAE', 'MAPE']].describe().loc[['mean','50%','std']].round(2))
    
    # 4. Stability Check
    mape_cv = (validation_metrics['MAPE'].std() / validation_metrics['MAPE'].mean()) * 100
    print(f"\n4. STABILITY:")
    print(f"- MAPE Coefficient of Variation: {mape_cv:.1f}%")
    
    # 5. Forecast Distribution
    preds = np.expm1(model_fit.predict())
    print(f"\n5. FORECAST DISTRIBUTION:")
    print(f"- Min: {preds.min():.1f} μg/m³")
    print(f"- Median: {np.median(preds):.1f} μg/m³")
    print(f"- Max: {preds.max():.1f} μg/m³")
    
    # 6. Recommendations
    print("\n6. RECOMMENDATIONS:")
    if len(sig_params) < len(model_fit.pvalues):
        print("- WARNING: Some parameters are insignificant - consider simplifying model")
    if jb_test[1] < 0.05:
        print("- WARNING: Residuals are not normally distributed - consider transformations")
    if mape_cv > 30:
        print("- WARNING: High error variability - monitor closely in production")
    else:
        print("- Model meets basic production criteria")

# Generate report
production_report(final_fit, validation_results[best_model_name])

# ======================
# VISUAL VALIDATION (FIXED)
# ======================

def plot_validation_results(forecast_data, model_name):
    """Plot actual vs predicted for all validation windows"""
    df = forecast_data[model_name]
    
    plt.figure(figsize=(15,6))
    
    # Convert pandas Series to numpy arrays before plotting
    dates = df['date'].values
    actual_values = df['actual'].values
    predicted_values = df['predicted'].values
    
    plt.plot(dates, actual_values, 'bo-', label='Actual', alpha=0.6, markersize=5)
    plt.plot(dates, predicted_values, 'ro--', label='Predicted', alpha=0.6, markersize=5)
    
    plt.title(f'Validation Results: {model_name}\nMAE: {validation_results[model_name]["MAE"].median():.1f} | MAPE: {validation_results[model_name]["MAPE"].median():.1f}%')
    plt.ylabel('PM2.5 (μg/m³)')
    plt.legend()
    plt.grid()
    plt.show()

# Plot best model validation
plot_validation_results(forecast_data, best_model_name)

# ======================
# SAVE PRODUCTION ARTIFACTS
# ======================

import joblib
import json
from datetime import datetime

# Save model
model_path = 'pm25_forecast_model.pkl'
joblib.dump(final_fit, model_path)

# Save metadata
metadata = {
    'model_type': 'SARIMAX',
    'order': list(best_order),
    'seasonal_order': list(best_seasonal),
    'last_training_date': str(data.index[-1]),
    'training_date': datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    'metrics': {
        'MAE': float(validation_results[best_model_name]['MAE'].median()),
        'MAPE': float(validation_results[best_model_name]['MAPE'].median()),
        'AIC': float(final_fit.aic),
        'BIC': float(final_fit.bic)
    },
    'data_characteristics': {
        'train_start': str(data.index[0]),
        'train_end': str(data.index[train_size-1]),
        'test_start': str(data.index[train_size]),
        'test_end': str(data.index[-1]),
        'n_observations': len(data)
    }
}

metadata_path = 'model_metadata.json'
with open(metadata_path, 'w') as f:
    json.dump(metadata, f, indent=2)

print("\nProduction artifacts saved:")
print(f"- Model: {model_path}")
print(f"- Metadata: {metadata_path}")
print("\nModel ready for production deployment!")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.seasonal import seasonal_decompose
from pmdarima import auto_arima
import scipy.stats as stats
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import warnings
from statsmodels.tsa.stattools import adfuller

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# Load and prepare data
df = pd.read_csv('c:/Users/kunal/OneDrive/CODE FOR LIFE/AQI/Analysis/datasets/kurla_daily.csv')
df['From Date'] = pd.to_datetime(df['From Date'])
df.set_index('From Date', inplace=True)
data = df['PM2.5_filled']

# ======================
# ENHANCED PREPROCESSING
# ======================

def enhanced_preprocessing(series, visualize=True):
    """Improved preprocessing pipeline"""
    # 1. Handle zeros (if any)
    series = series.replace(0, np.nan).interpolate()
    
    # 2. Log transformation (add small constant to handle near-zero values)
    transformed = np.log1p(series)
    
    # 3. Remove outliers using IQR
    Q1 = transformed.quantile(0.25)
    Q3 = transformed.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    transformed = transformed[(transformed >= lower_bound) & (transformed <= upper_bound)]
    
    if visualize:
        fig, ax = plt.subplots(2, 1, figsize=(12, 8))
        series.plot(ax=ax[0], title='Original Series')
        pd.Series(transformed, index=series.index).plot(ax=ax[1], title='Transformed Series')
        plt.tight_layout()
        plt.show()
    
    return transformed

data_transformed = enhanced_preprocessing(data)
train_size = int(len(data_transformed) * 0.8)  # Increased training size
train, test = data_transformed.iloc[:train_size], data_transformed.iloc[train_size:]

# ======================
# SEASONALITY ANALYSIS
# ======================

def analyze_seasonality(series, period=30):
    """Enhanced seasonality analysis"""
    # Decomposition
    result = seasonal_decompose(series, period=period, model='additive')
    
    # Plot decomposition
    fig = result.plot()
    fig.set_size_inches(12, 8)
    plt.tight_layout()
    plt.show()
    
    # ACF/PACF plots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    plot_acf(series, lags=60, ax=ax1)
    plot_pacf(series, lags=30, ax=ax2)
    plt.tight_layout()
    plt.show()
    
    return result

seasonal_result = analyze_seasonality(train)

# ======================
# AUTO MODEL SELECTION
# ======================

def auto_select_model(train_series, seasonal_period=30):
    """Automatically select best SARIMAX parameters"""
    auto_model = auto_arima(
        train_series,
        start_p=0, max_p=3,
        start_q=0, max_q=3,
        d=None, max_d=2,
        seasonal=True,
        m=seasonal_period,
        start_P=0, max_P=2,
        start_Q=0, max_Q=2,
        D=None, max_D=1,
        trace=True,
        error_action='ignore',
        suppress_warnings=True,
        stepwise=True,
        information_criterion='aic'
    )
    
    print("\nAuto-selected model summary:")
    print(auto_model.summary())
    
    return (auto_model.order, auto_model.seasonal_order)

auto_order, auto_seasonal = auto_select_model(train)

# ======================
# ENHANCED VALIDATION FRAMEWORK
# ======================

def enhanced_walk_forward_validation(data, train_size, order, seasonal_order, steps=7):
    """More robust validation with additional metrics"""
    test = data.iloc[train_size:]
    n_windows = len(test) // steps
    metrics = []
    forecasts = []
    
    for i in range(n_windows):
        train_window = data.iloc[:train_size + i*steps]
        
        try:
            model = SARIMAX(
                train_window,
                order=order,
                seasonal_order=seasonal_order,
                enforce_stationarity=False,
                enforce_invertibility=False,
                initialization='approximate_diffuse'
            )
            
            model_fit = model.fit(disp=False, maxiter=200, method='powell')
            
            forecast = model_fit.get_forecast(steps=steps)
            pred = np.expm1(forecast.predicted_mean)
            actual = np.expm1(test.iloc[i*steps:(i+1)*steps])
            
            # Calculate multiple metrics
            mae = mean_absolute_error(actual, pred)
            rmse = np.sqrt(mean_squared_error(actual, pred))
            mape = np.mean(np.abs((actual - pred)/actual)) * 100
            smape = 100 * np.mean(2 * np.abs(pred - actual) / (np.abs(pred) + np.abs(actual)))
            
            metrics.append({
                'Window': i+1,
                'MAE': mae,
                'RMSE': rmse,
                'MAPE': mape,
                'sMAPE': smape,
                'AIC': model_fit.aic,
                'BIC': model_fit.bic,
                'HQIC': model_fit.hqic
            })
            
            forecasts.append({
                'date': test.index[i*steps:(i+1)*steps],
                'actual': actual.values,
                'predicted': pred.values
            })
            
        except Exception as e:
            print(f"Window {i+1} failed: {str(e)}")
            continue
            
    return pd.DataFrame(metrics), pd.concat([pd.DataFrame(f) for f in forecasts])

# ======================
# MODEL COMPARISON
# ======================

models_to_test = {
    "Auto-selected Model": (auto_order, auto_seasonal),
    "Robust (1,1,1)x(0,1,1,30)": ((1,1,1), (0,1,1,30)),
    "Simplified (1,1,0)x(0,1,0,30)": ((1,1,0), (0,1,0,30)),
    "Complex (2,1,2)x(1,1,1,30)": ((2,1,2), (1,1,1,30))
}

validation_results = {}
forecast_data = {}

for name, params in models_to_test.items():
    print(f"\n=== Validating {name} ===")
    res, forecasts = enhanced_walk_forward_validation(data_transformed, train_size, *params)
    validation_results[name] = res
    forecast_data[name] = forecasts
    print(res[['MAE', 'MAPE', 'sMAPE']].describe())

# ======================
# SELECT BEST MODEL
# ======================

# Use sMAPE for model selection as it's more robust to outliers
model_scores = {name: res['sMAPE'].median() for name, res in validation_results.items()}
best_model_name = min(model_scores, key=model_scores.get)
best_order, best_seasonal = models_to_test[best_model_name]

print(f"\n★★★ Best Model: {best_model_name} ★★★")
print(f"Median sMAPE: {model_scores[best_model_name]:.2f}%")
print(f"Median MAE: {validation_results[best_model_name]['MAE'].median():.2f}")
print(f"Order: {best_order}, Seasonal Order: {best_seasonal}")

# ======================
# FINAL MODEL TRAINING
# ======================

# Train final model on full training data with best parameters
final_model = SARIMAX(
    data_transformed,
    order=best_order,
    seasonal_order=best_seasonal,
    enforce_stationarity=False,
    enforce_invertibility=False,
    initialization='approximate_diffuse'
)

final_fit = final_model.fit(disp=True, maxiter=300, method='powell')

# ======================
# IMPROVED FORECAST VISUALIZATION
# ======================

def enhanced_forecast_plot(model_fit, history, steps=30, ci=True):
    """Generate production forecast with enhanced visualization"""
    # Get forecast
    forecast = model_fit.get_forecast(steps=steps)
    pred = pd.Series(np.expm1(forecast.predicted_mean), name='forecast')
    conf_int = np.expm1(forecast.conf_int())
    
    # Create output dataframe
    forecast_dates = pd.date_range(start=history.index[-1] + pd.Timedelta(days=1), periods=steps)
    result = pd.DataFrame({
        'date': forecast_dates,
        'PM2.5_forecast': pred.values,
        'lower_CI': conf_int.iloc[:,0],
        'upper_CI': conf_int.iloc[:,1]
    }).set_index('date')
    
    # Plot
    plt.figure(figsize=(15, 7))
    
    # Convert history to numpy arrays for plotting
    hist_dates = history.index[-180:].to_numpy()
    hist_values = history.values[-180:]
    
    plt.plot(hist_dates, hist_values, 
            'b-', label='Historical PM2.5', alpha=0.8, linewidth=2)
    
    # Convert forecast to numpy arrays
    fcst_dates = result.index.to_numpy()
    fcst_values = result['PM2.5_forecast'].to_numpy()
    lower_ci = result['lower_CI'].to_numpy()
    upper_ci = result['upper_CI'].to_numpy()
    
    plt.plot(fcst_dates, fcst_values, 
            'r--', label='Forecast', linewidth=2, marker='o', markersize=5)
    
    if ci:
        plt.fill_between(fcst_dates, lower_ci, upper_ci, 
                       color='pink', alpha=0.3, label='95% CI')
    
    plt.title(f'PM2.5 {steps}-Day Forecast | {best_model_name}\n'
             f"sMAPE: {model_scores[best_model_name]:.2f}% | "
             f"MAE: {validation_results[best_model_name]['MAE'].median():.2f}", pad=20)
    plt.ylabel('PM2.5 (μg/m³)')
    plt.xlabel('Date')
    plt.legend(loc='upper left')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
    
    return result

# Generate forecast plot
production_forecast = enhanced_forecast_plot(final_fit, data, steps=30)

# ======================
# ENHANCED PRODUCTION REPORT
# ======================

def enhanced_production_report(model_fit, validation_metrics):
    """Comprehensive deployment checklist with more diagnostics"""
    print("\n" + "="*60)
    print("ENHANCED PRODUCTION READINESS REPORT".center(60))
    print("="*60)
    
    # 1. Model Summary
    print("\n1. MODEL SUMMARY:")
    print(f"SARIMAX{model_fit.model.order}x{model_fit.model.seasonal_order}")
    print(f"\nModel AIC: {model_fit.aic:.2f}, BIC: {model_fit.bic:.2f}")
    
    # 2. Parameter Significance
    print("\n2. PARAMETER SIGNIFICANCE (p < 0.05):")
    sig_params = model_fit.pvalues[model_fit.pvalues < 0.05]
    if len(sig_params) > 0:
        print(sig_params.to_string())
    else:
        print("No significant parameters at p < 0.05 level")
    
    # 3. Residual Diagnostics
    print("\n3. RESIDUAL DIAGNOSTICS:")
    residuals = model_fit.resid.dropna()
    
    # Normality tests
    jb_test = stats.jarque_bera(residuals)
    shapiro_test = stats.shapiro(residuals)
    
    # Stationarity tests
    adf_test = adfuller(residuals)
    
    # Autocorrelation
    lb_test = acorr_ljungbox(residuals, lags=[10], return_df=True)
    
    print(f"- Normality (JB test): statistic={jb_test[0]:.2f}, p-value={jb_test[1]:.4f}")
    print(f"- Normality (Shapiro-Wilk): statistic={shapiro_test[0]:.2f}, p-value={shapiro_test[1]:.4f}")
    print(f"- Stationarity (ADF): p-value={adf_test[1]:.4f}")
    print(f"- Autocorrelation (Ljung-Box): p-value={lb_test['lb_pvalue'].iloc[0]:.4f}")
    
    # 4. Validation Performance
    print("\n4. VALIDATION PERFORMANCE:")
    print(validation_metrics[['MAE', 'RMSE', 'MAPE', 'sMAPE']].describe().loc[['mean','50%','std']].round(2))
    
    # 5. Stability Analysis
    mape_cv = (validation_metrics['MAPE'].std() / validation_metrics['MAPE'].mean()) * 100
    print(f"\n5. STABILITY ANALYSIS:")
    print(f"- MAPE Coefficient of Variation: {mape_cv:.1f}%")
    print(f"- MAE Range: {validation_metrics['MAE'].min():.2f} to {validation_metrics['MAE'].max():.2f}")
    
    # 6. Forecast Characteristics
    preds = np.expm1(model_fit.predict())
    print(f"\n6. FORECAST CHARACTERISTICS:")
    print(f"- Forecast Range: {preds.min():.1f} to {preds.max():.1f} μg/m³")
    print(f"- Forecast Mean: {preds.mean():.1f} μg/m³")
    print(f"- Forecast Std Dev: {preds.std():.1f} μg/m³")
    
    # 7. Recommendations
    print("\n7. RECOMMENDATIONS:")
    if len(sig_params) < len(model_fit.pvalues)/2:
        print("- WARNING: Many parameters are insignificant - consider model simplification")
    if jb_test[1] < 0.05 or shapiro_test[1] < 0.05:
        print("- WARNING: Residuals are not normally distributed - consider transformations")
    if mape_cv > 30:
        print("- WARNING: High error variability - consider ensemble methods")
    if validation_metrics['sMAPE'].median() > 15:
        print("- WARNING: sMAPE > 15% - model may need additional features or different approach")
    else:
        print("- Model meets production criteria for deployment")

# Generate enhanced report
enhanced_production_report(final_fit, validation_results[best_model_name])

# ======================
# SAVE PRODUCTION ARTIFACTS
# ======================

import joblib
import json
from datetime import datetime

# Save model
model_path = 'improved_pm25_forecast_model.pkl'
joblib.dump(final_fit, model_path)

# Save metadata with more details
metadata = {
    'model_type': 'SARIMAX',
    'order': list(best_order),
    'seasonal_order': list(best_seasonal),
    'last_training_date': str(data.index[-1]),
    'training_date': datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    'metrics': {
        'MAE': float(validation_results[best_model_name]['MAE'].median()),
        'RMSE': float(validation_results[best_model_name]['RMSE'].median()),
        'MAPE': float(validation_results[best_model_name]['MAPE'].median()),
        'sMAPE': float(validation_results[best_model_name]['sMAPE'].median()),
        'AIC': float(final_fit.aic),
        'BIC': float(final_fit.bic),
        'HQIC': float(final_fit.hqic)
    },
    'data_characteristics': {
        'train_start': str(data.index[0]),
        'train_end': str(data.index[train_size-1]),
        'test_start': str(data.index[train_size]),
        'test_end': str(data.index[-1]),
        'n_observations': len(data),
        'mean_pm25': float(data.mean()),
        'std_pm25': float(data.std())
    },
    'validation_windows': len(validation_results[best_model_name]),
    'recommendations': "Monitor performance weekly and retrain monthly"
}

metadata_path = 'improved_model_metadata.json'
with open(metadata_path, 'w') as f:
    json.dump(metadata, f, indent=2)

print("\nProduction artifacts saved:")
print(f"- Model: {model_path}")
print(f"- Metadata: {metadata_path}")
print("\n★★★ Model ready for production deployment! ★★★")
```
